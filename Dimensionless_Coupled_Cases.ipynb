# Dimensionless Ambipolar Transport Equation

## Objective

The objective of this notebook is to derive the dimensionless form of the one-dimensional ambipolar transport equation including thermodiffusion (Soret effect).

The governing equation describes carrier transport due to diffusion, drift, and thermally driven migration. The temperature field is treated as an externally known quantity and influences the carrier transport through the thermodiffusion flux.

---

# 1. Dimensional Governing Equation

The dimensional ambipolar transport equation is

$$
\frac{\partial p}{\partial t}
=
D\frac{\partial^2p}{\partial x^2}
-
v\frac{\partial p}{\partial x}
+
DS
\frac{\partial}{\partial x}
\left(
p
\frac{\partial T}{\partial x}
\right).
$$

The three terms on the right-hand side represent

- Diffusion,
- Drift,
- Thermodiffusion (Soret effect).

---

# 2. Dimensionless Variables

Using the characteristic scales,

Dimensionless spatial coordinate

$$
X=\frac{x}{L},
$$

Dimensionless time

$$
\tau=\frac{\alpha t}{L^2},
$$

Dimensionless carrier concentration

$$
P=\frac{p}{p_0},
$$

Dimensionless temperature

$$
\theta=\frac{T-T_0}{\Delta T}.
$$

Hence,

$$
x=LX,
$$

$$
t=\frac{L^2}{\alpha}\tau,
$$

$$
p=p_0P,
$$

and

$$
T=T_0+\Delta T\,\theta.
$$

---

# 3. Transformation of the Derivatives

Using the chain rule,

Time derivative

$$
\frac{\partial p}{\partial t}
=
\frac{p_0\alpha}{L^2}
\frac{\partial P}{\partial\tau}.
$$

First spatial derivative

$$
\frac{\partial p}{\partial x}
=
\frac{p_0}{L}
\frac{\partial P}{\partial X}.
$$

Second spatial derivative

$$
\frac{\partial^2p}{\partial x^2}
=
\frac{p_0}{L^2}
\frac{\partial^2P}{\partial X^2}.
$$

Temperature gradient

$$
\frac{\partial T}{\partial x}
=
\frac{\Delta T}{L}
\frac{\partial\theta}{\partial X}.
$$

---

# 4. Transformation of the Thermodiffusion Term

The thermodiffusion term is

$$
DS
\frac{\partial}{\partial x}
\left(
p
\frac{\partial T}{\partial x}
\right).
$$

Substituting the dimensionless variables,

$$
=
DS
\frac{\partial}{\partial x}
\left(
p_0P
\cdot
\frac{\Delta T}{L}
\frac{\partial\theta}{\partial X}
\right).
$$

Taking the constants outside,

$$
=
DS
\frac{p_0\Delta T}{L}
\frac{\partial}{\partial x}
\left(
P
\frac{\partial\theta}{\partial X}
\right).
$$

Using

$$
\frac{\partial}{\partial x}
=
\frac{1}{L}
\frac{\partial}{\partial X},
$$

gives

$$
DS
\frac{\partial}{\partial x}
\left(
p
\frac{\partial T}{\partial x}
\right)
=
DS
\frac{p_0\Delta T}{L^2}
\frac{\partial}{\partial X}
\left(
P
\frac{\partial\theta}{\partial X}
\right).
$$

---

# 5. Dimensionless Governing Equation

Substituting all transformed terms into the dimensional equation gives

$$
\frac{p_0\alpha}{L^2}
\frac{\partial P}{\partial\tau}
=
D
\frac{p_0}{L^2}
\frac{\partial^2P}{\partial X^2}
-
v
\frac{p_0}{L}
\frac{\partial P}{\partial X}
+
DS
\frac{p_0\Delta T}{L^2}
\frac{\partial}{\partial X}
\left(
P
\frac{\partial\theta}{\partial X}
\right).
$$

Dividing throughout by

$$
\frac{p_0\alpha}{L^2},
$$

gives

$$
\frac{\partial P}{\partial\tau}
=
\frac{D}{\alpha}
\frac{\partial^2P}{\partial X^2}
-
\frac{vL}{\alpha}
\frac{\partial P}{\partial X}
+
\frac{DS\Delta T}{\alpha}
\frac{\partial}{\partial X}
\left(
P
\frac{\partial\theta}{\partial X}
\right).
$$

---

# 6. Dimensionless Parameters

Define the dimensionless parameters

$$
Le=\frac{D}{\alpha},
$$

$$
Pe=\frac{vL}{\alpha},
$$

and

$$
\boxed{
\lambda
=
\frac{DS\Delta T}{\alpha}
}
$$

where

- \(Le\) is the Lewis number,
- \(Pe\) is the Péclet number,
- \(\lambda\) is the thermodiffusion coupling parameter.

---

# 7. Final Dimensionless Ambipolar Transport Equation

The final dimensionless governing equation is

$$
\boxed{
\frac{\partial P}{\partial\tau}
=
Le
\frac{\partial^2P}{\partial X^2}
-
Pe
\frac{\partial P}{\partial X}
+
\lambda
\frac{\partial}{\partial X}
\left(
P
\frac{\partial\theta}{\partial X}
\right)
}
$$

where

$$
\lambda=\frac{DS\Delta T}{\alpha}.
$$

The thermodiffusion term is written in conservative (divergence) form because it directly represents the thermodiffusion flux and preserves the conservation form of the governing equation.

---

# 8. Expanded Form (Equivalent)

Applying the product rule,

$$
\frac{\partial}{\partial X}
\left(
P
\frac{\partial\theta}{\partial X}
\right)
=
P
\frac{\partial^2\theta}{\partial X^2}
+
\frac{\partial P}{\partial X}
\frac{\partial\theta}{\partial X}.
$$

Therefore, the governing equation can also be written as

$$
\boxed{
\frac{\partial P}{\partial\tau}
=
Le
\frac{\partial^2P}{\partial X^2}
-
Pe
\frac{\partial P}{\partial X}
+
\lambda
\left(
P
\frac{\partial^2\theta}{\partial X^2}
+
\frac{\partial P}{\partial X}
\frac{\partial\theta}{\partial X}
\right)
}
$$

Both equations are mathematically equivalent.

However, throughout this work, the conservative (divergence) form is adopted because

- it directly represents the thermodiffusion flux,
- it preserves the conservative form of the transport equation,
- it is the form commonly used in transport theory,
- it is naturally compatible with finite-volume formulations,
- it is more conveniently implemented in FiPy.

Therefore, the conservative form is used for all subsequent derivations and numerical implementation.

---

# 9. Summary of Dimensionless Parameters

| Symbol | Definition | Physical Meaning |
|---------|------------|------------------|
| $Le$ | $\dfrac{D}{\alpha}$ | Lewis number |
| $Pe$ | $\dfrac{vL}{\alpha}$ | Péclet number |
| $\lambda$ | $\dfrac{DS\Delta T}{\alpha}$ | Thermodiffusion coupling parameter |

---

# Conclusion

The one-dimensional ambipolar transport equation has been successfully transformed into a dimensionless form using the characteristic length, time, concentration, and temperature scales. The resulting formulation introduces the dimensionless thermodiffusion coupling parameter, $\lambda$, which quantifies the influence of temperature gradients on carrier transport. The governing equation is expressed in conservative (divergence) form to preserve the flux-based representation of thermodiffusion and to facilitate finite-volume discretization. The equivalent expanded form is also presented for completeness. This dimensionless formulation provides the mathematical foundation for the numerical implementation of the ambipolar transport solver in the subsequent notebook.

# Trivial Initial Condition Test

To verify the numerical implementation, a trivial initial-condition test is performed before considering more complex initial profiles.

The initial temperature and carrier concentration are chosen as linear functions:

$$
\theta(X,0)=X
$$

$$
P(X,0)=X
$$

A linear profile is selected because it has a constant first derivative and zero second derivative, making it one of the simplest profiles for validating the numerical solver.

All governing equations, boundary conditions, and dimensionless parameters remain unchanged. Only the initial conditions are modified.

```python
import numpy as np
import matplotlib.pyplot as plt
from fipy import *

# ==========================================================
# Dimensionless One-Way Coupled Heat–Ambipolar Transport Model
#
# Heat:
#     ∂θ/∂τ = ∂²θ/∂X² + Q*
#
# Transport:
#     ∂P/∂τ = Le∂²P/∂X²
#              - Pe∂P/∂X
#              + λ∇·(P∇θ)
#
# Trivial Test Case:
# Initial conditions
#
#     θ(X,0) = X
#     P(X,0) = X
#
# ==========================================================

# ----------------------------------------------------------
# Mesh
# ----------------------------------------------------------

L = 1.0
nx = 100
dx = L / nx

mesh = Grid1D(nx=nx, dx=dx)

# Cell-center coordinates
x = mesh.cellCenters[0]

# ----------------------------------------------------------
# Dimensionless Parameters
# ----------------------------------------------------------

Q_star = 1.0

Le = 1.0
Pe = 2.0
lambda_ = 0.5

dt = 1e-3
max_steps = 10000
tolerance = 1e-8

# ==========================================================
# HEAT EQUATION
# ==========================================================

theta = CellVariable(
    mesh=mesh,
    value=x,
    name="Temperature"
)

theta.constrain(0.0, mesh.facesLeft)
theta.constrain(1.0, mesh.facesRight)

heatEq = (
    TransientTerm()
    ==
    DiffusionTerm(coeff=1.0)
    + Q_star
)

for step in range(max_steps):

    old = theta.value.copy()

    heatEq.solve(var=theta, dt=dt)

    if np.max(np.abs(theta.value - old)) < tolerance:
        print(f"Heat equation converged after {step+1} steps")
        break

# ==========================================================
# UNCOUPLED TRANSPORT
# ==========================================================

P_uncoupled = CellVariable(
    mesh=mesh,
    value=x,
    name="Uncoupled"
)

P_uncoupled.constrain(0.0, mesh.facesLeft)
P_uncoupled.constrain(1.0, mesh.facesRight)

transportEq = (
    TransientTerm()
    ==
    DiffusionTerm(coeff=Le)
    -
    PowerLawConvectionTerm(coeff=(Pe,))
)

for step in range(max_steps):

    old = P_uncoupled.value.copy()

    transportEq.solve(var=P_uncoupled, dt=dt)

    if np.max(np.abs(P_uncoupled.value - old)) < tolerance:
        print(f"Uncoupled transport converged after {step+1} steps")
        break

# ==========================================================
# COUPLED TRANSPORT
# ==========================================================

P_coupled = CellVariable(
    mesh=mesh,
    value=x,
    name="Coupled"
)

P_coupled.constrain(0.0, mesh.facesLeft)
P_coupled.constrain(1.0, mesh.facesRight)

# Thermodiffusion velocity
thermalVelocity = lambda_ * theta.faceGrad

coupledEq = (
    TransientTerm()
    ==
    DiffusionTerm(coeff=Le)
    -
    PowerLawConvectionTerm(coeff=(Pe,))
    +
    PowerLawConvectionTerm(coeff=thermalVelocity)
)

for step in range(max_steps):

    old = P_coupled.value.copy()

    coupledEq.solve(var=P_coupled, dt=dt)

    if np.max(np.abs(P_coupled.value - old)) < tolerance:
        print(f"Coupled transport converged after {step+1} steps")
        break

# ==========================================================
# COMPARISON
# ==========================================================

difference = np.abs(P_coupled.value - P_uncoupled.value)

max_difference = difference.max()
l2_difference = np.sqrt(np.mean(difference**2))

print("\nMaximum Difference :", max_difference)
print("L2 Difference      :", l2_difference)

# ==========================================================
# PLOTS
# ==========================================================

x_plot = np.array(mesh.cellCenters[0])

plt.figure(figsize=(8, 5))
plt.plot(x_plot, theta.value, linewidth=2)
plt.grid(True)
plt.xlabel("Dimensionless Position")
plt.ylabel("Temperature")
plt.title("Dimensionless Temperature")
plt.show()

# Temperature gradient (defined on faces)

x_faces = np.array(mesh.faceCenters[0])
gradT = np.array(theta.faceGrad[0])

plt.figure(figsize=(8, 5))
plt.plot(x_faces, gradT, linewidth=2)
plt.grid(True)
plt.xlabel("Dimensionless Position")
plt.ylabel("Temperature Gradient")
plt.title("Dimensionless Temperature Gradient")
plt.show()

plt.figure(figsize=(8, 5))

plt.plot(
    x_plot,
    P_uncoupled.value,
    linewidth=2,
    label="Uncoupled"
)

plt.plot(
    x_plot,
    P_coupled.value,
    "--",
    linewidth=2,
    label="Coupled"
)

plt.grid(True)
plt.legend()
plt.xlabel("Dimensionless Position")
plt.ylabel("Carrier Concentration")
plt.title("Carrier Concentration")
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(x_plot, difference, linewidth=2)
plt.grid(True)
plt.xlabel("Dimensionless Position")
plt.ylabel("Difference")
plt.title("Difference Between Coupled and Uncoupled")
plt.show()

# ==========================================================
# SUMMARY
# ==========================================================

print("\n==============================")
print("COUPLED MODEL SUMMARY")
print("==============================")

print(f"Heat Source (Q*)        : {Q_star}")
print(f"Diffusion Ratio (Le)    : {Le}")
print(f"Peclet Number (Pe)      : {Pe}")
print(f"Coupling Parameter (λ)  : {lambda_}")

print(f"\nMaximum Difference : {max_difference:.6e}")
print(f"L2 Difference      : {l2_difference:.6e}")
```

# Results and Discussion

The objective of this test was to verify the numerical implementation using the simplest possible initial conditions,

$$
\theta(X,0)=X,
$$

$$
P(X,0)=X.
$$

Although both fields are initialized as straight lines, the governing equations determine how these profiles evolve with time until steady state. The obtained results are discussed below.

---

# 1. Temperature Distribution

The initial temperature profile is a straight line,

$$
\theta(X,0)=X.
$$

However, the temperature does not remain linear during the simulation. Instead, it gradually evolves into a curved profile before reaching steady state.

This behaviour is expected because the dimensionless heat equation being solved is

$$
\frac{\partial \theta}{\partial \tau}
=
\frac{\partial^2\theta}{\partial X^2}
+
Q^*,
$$

where

$$
Q^*=1.
$$

The constant source term represents uniform heat generation throughout the entire domain.

Initially,

$$
\frac{\partial^2\theta}{\partial X^2}=0,
$$

since a straight line has zero curvature.

However, because

$$
Q^*>0,
$$

the governing equation becomes

$$
\frac{\partial\theta}{\partial\tau}>0,
$$

which means the temperature field cannot remain linear. As time progresses, heat is continuously generated inside the domain, causing the profile to bend until diffusion balances the heat generation.

At steady state,

$$
\frac{\partial\theta}{\partial\tau}=0,
$$

therefore

$$
\frac{d^2\theta}{dX^2}
=
-
Q^*.
$$

Since

$$
Q^*=1,
$$

the steady-state equation becomes

$$
\frac{d^2\theta}{dX^2}=-1.
$$

A function whose second derivative is constant is a quadratic (parabolic) function. Therefore, the final temperature profile is expected to be curved rather than linear, exactly as observed in the numerical solution.

---

# 2. Temperature Gradient

The temperature gradient represents the slope of the temperature profile,

$$
\frac{d\theta}{dX}.
$$

Initially, the temperature profile is linear, so the gradient is constant throughout the domain.

As the temperature profile gradually changes from a straight line to a parabola due to internal heat generation, its gradient also changes.

Since the derivative of a quadratic function is a linear function, the temperature gradient decreases almost linearly from the left boundary to the right boundary.

Therefore, the nearly straight decreasing gradient observed in the numerical solution is the expected physical behaviour.

---

# 3. Carrier Concentration

Both the uncoupled and coupled carrier transport simulations start from the same initial condition,

$$
P(X,0)=X.
$$

The uncoupled transport equation includes only diffusion and convection.

The coupled transport equation additionally contains the thermodiffusion contribution,

$$
\lambda\nabla\cdot(P\nabla\theta),
$$

which depends directly on the computed temperature gradient.

Since the temperature field is no longer perfectly linear after solving the heat equation, its spatial gradient varies throughout the domain. Consequently, the thermodiffusion term modifies the carrier transport, producing a carrier concentration profile that differs slightly from the uncoupled solution.

The coupled solution remains above the uncoupled solution over most of the domain, indicating that the temperature gradient assists carrier transport in the positive direction for the chosen parameter values.

---

# 4. Difference Between Coupled and Uncoupled Solutions

The difference plot represents

$$
|P_{\mathrm{coupled}}-P_{\mathrm{uncoupled}}|.
$$

At both boundaries, the carrier concentration is prescribed through Dirichlet boundary conditions,

$$
P(0)=0,
$$

$$
P(1)=1.
$$

Since both simulations satisfy the same boundary conditions, the difference between them is nearly zero near the boundaries.

Moving away from the boundaries, the influence of the thermodiffusion term gradually increases because the carrier transport is affected by the temperature gradient.

The maximum difference occurs approximately in the interior of the domain, where the cumulative effect of thermodiffusion is strongest before both solutions are constrained again by the right boundary condition.

---

# Overall Interpretation

This verification demonstrates that the numerical implementation behaves consistently.

Although both temperature and carrier concentration were initialized using simple linear profiles, the governing equations determine the final steady-state solution.

The heat equation transforms the initial linear temperature profile into the physically expected nonlinear steady-state distribution because of the internal heat source.

The resulting temperature gradient then influences the carrier transport through the thermodiffusion term, producing a measurable difference between the coupled and uncoupled carrier concentration profiles.

The obtained numerical results are therefore consistent with the mathematical behaviour of the governing equations and provide confidence in the correctness of the one-way coupled implementation before investigating more complex initial conditions.
